```python
# global variable
balance= 1000
def withdraw(amount):
    global balance 
    if amount<=balance:
        balance-=amount
        print(f"Withdrawn: {amount}, Remaining Balance: {balance}" )
    else:
        print("Insufficient Balance!..")
def check_balance():
    print(f"Current Balance: {balance}")
withdraw(200)
check_balance()
withdraw(900)
```

```python
def add(*num):
    sum=0
    for n in num:
        sum=sum+n
    print("sum of the number is",sum)
add(3,5)
add(4,5,6,7)
add(1,2,3,5,6)
```

```python
def make_sentence(*words):
    sentence= ''.join(words).capitalize()+'.'
    print("Generated Sentence:",sentence)

make_sentence("learning ","python0","is","fun")
make_sentence("this","is","another","example")
make_sentence("awesome")
```

```python
def  show_cart(**items):
    print("Shopping Cart Summary:")
    total=0
    for item,price in items.items():
        print(f"- {item}:{price}")
        total+= price
    print(f"total Amount: {total}")
    print("."*30)
show_cart(apple=30,milk=50,bread= 25)
show_cart(pen=10,notebook=40,eraser=5,ruler=15)
show_cart(banana=40,apple=105,pineapple=60,chicku=80)
```
