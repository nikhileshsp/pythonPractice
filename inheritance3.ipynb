```python
class Person:
    def person_info(self,name,age):
        print('Inside Person class')
        print('name:',name,'age:',age)

class Company:
    def company_info(self,company_name,location):
        print('Insede Company class')
        print('name:',company_name,'location:',location)

class Employee(Person,Company):
    def employee_info(self,salary,skill):
        print('salary:',salary,'skill:',skill)

emp=Employee()
emp.person_info('madhu',28)
emp.company_info('ibm','pune')
emp.employee_info(92000,'Python')
```

```python
class Vehicle:
    def __init__(self,companyname,year):
        self.companyname=companyname
        self.year=year
    def Vehicle_info(self):
        print('My Vehicle Details are',self.companyname,self.year)

class Car(Vehicle):
    def __init__(self,companyname,year,carcolor,purchasedate):
        super().__init__(companyname,year)  # used super() to keep the inheritance
        self.carcolor=carcolor
        self.purchasedate=purchasedate
    def car_info(self):
        print('My Car Details is ',self.carcolor,self.purchasedate)

class LuxuaryCar(Car):
    def __init__(self,companyname,year,carcolor,purchasedate,carname):
        Car.__init__(self,companyname,year,carcolor,purchasedate)
        self.carname=carname
    def luxuary_info(self):
        print('My Car Is ',self.carname)

car1= LuxuaryCar("Bayerische Motoren Werke AG",2019,"Red","17th jun 2022","BMW Alpina")
car1.Vehicle_info()
car1.car_info()
car1.luxuary_info()
    
```

```python
#Hybrid Inheritance with Parameterized Constructor

class A:
    def __init__(self,name,id):
        self.name=name
        self.id= id
    def customer_info1(self):
        print("customer name and id is ",self.name,self.id)

class B(A):
    def __init__(self,name,id,mname,area):
        A.__init__(self,name,id)
        self.mname=mname
        self.area= area
    def customer_info2(self):
        print("customer mname and area is",self.mname,self.area)

class C(A):
    def __init__(self,name,id,lname,city):
        A.__init__(self,name,id)
        self.lname=lname
        self.city= city
    def customer_info3(self):
        print("customer lname and city is",self.lname,self.city)

class D(A):
    def __init__(self,name,id,pincode):
        A.__init__(self,name,id)
        self.pincode=pincode
    def customer_info4(self):
        print("customer pincode is",self.pincode)

class E(B,C):
    def __init__(self,name,id,mname,area,lname,city,state):
        B.__init__(self,name,id,mname,area)
        C.__init__(self,name,id,lname,city)
        self.state=state
    def customer_info5(self):
        print("state :",self.state)
        
        

class F(D,E):
    def __init__(self,name,id,mname,area,lname,city,state,pincode,country):
        D.__init__(self,name,id,pincode)
        E.__init__(self,name,id,mname,area,lname,city,state)
        self.country=country
    def customer_info6(self):
        print("country: ",self.country)
        
        
cust = F("Nikhilesh", 1, "Subhash", "Hingna", "Patel", "Nagpur", "Maharashtra", "440016", "India")

cust.customer_info1()
cust.customer_info2()
cust.customer_info3()
cust.customer_info4()
cust.customer_info5()
cust.customer_info6()

    
    
```

```python
#Hybrid Inheritance with Default Constructor

class A:
    def classa(self,name,id):
        print(name,id)
        
class B(A):
    def classb(self,mname,area):
        print(mname,area)
        
class C(A):
    def classc(self,lname,city):
        print(lname,city)
        
class D(A):
    def classd(self,pincode):
        print(pincode)
        
class E(B,C):
    def classe(self,state):
        print(state)
        
class F(D,E):
    def classf(self,country):
        print(country)
#creating class object
f1=F()
f1.classa("Rashmi","E01")
f1.classb("Manju","Gayatri Nagar")
f1.classc("Joshi","Nagpur")
f1.classd("440022")
f1.classe("Maharashtra")
f1.classf("India")
```

```python
#Hybrid Inheritance with Non-Parameterized Constructor

class A:
    def __init__(self):
        print("Printing Class A Objects First Name & Id")
    def classa(self):
        print("Rashmi","E01")
        
class B(A):
    def __init__(self):
        print("Printing Class B Objects Middle Name & Area")
    def classb(self):
        print("Manju","Gayatri Nagar")
        
class C(A):
    def __init__(self):
        print("Printing Class C Objects Last Name & City")
    def classc(self):
        print("Joshi","Nagpur")
        
class D(A):
    def __init__(self):
        print("Printing Class D Objects Pincode")
    def classd(self):
        print("440022")
class E(B,C):
    def __init__(self):
        print("Printing Class E Objects State")
    def classe(self):
        print("Maharashtra")
        
class F(D,E):
    def __init__(self):
        print("Printing Class F Objects Country")
    def classf(self):
        print("India")
#creating class object
f1=F()
f1.classa()
f1.classb()
f1.classc()
f1.classd()
f1.classe()
f1.classf()
```
