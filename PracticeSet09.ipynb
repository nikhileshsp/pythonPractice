```python
# 1. Write a list comprehension to create a list of squares of numbers from 1 to 10.

output_list_comp= [i **2 for i in range(2,11)]

print(output_list_comp )
```

```python
# 2. Use a list comprehension to extract all the even numbers from a given list of integers.

output_list_cmp= [2,3,4,5,6,6,9,9,8,8,7]
set_using_comp={var for var in output_list_cmp if var%2==0 }
print(set_using_comp)

numbers = [1, 2, 3, 4, 5, 6,6, 7, 8, 9, 10]
evens = [x for x in numbers if x % 2 == 0]
print("Even numbers:", evens)
```

```python
# 3. Create a list comprehension to find the common elements between two lists.

list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]
common_elements = [x for x in list1 if x in list2]
print("Common elements:", common_elements)
```

```python
# 4. Write a list comprehension to conv?ert a list of strings to uppercase.
# Dictionary Comprehensions:

words = ["apple", "banana", "cherry"]
upper_words = [word.upper() for word in words]
print("Uppercase words:", upper_words)
```

```python
# 5. Create a dictionary comprehension to convert a list of names into a dictionary where the
# keys are names and the values are their lengths.

names = ["aman", "ajay", "nikita"]
name_lengths = {name: len(name) for name in names}
print("Name lengths:", name_lengths)
```

```python
# 6. Use a dictionary comprehension to count the frequency of words in a given list of words.

word_list = ["apple", "banana", "apple", "orange", "banana", "banana"]
word_freq = {word: word_list.count(word) for word in set(word_list)}
print("Word frequencies:", word_freq)
```

```python
# 7. Write a dictionary comprehension to swap the keys and values of a dictionary.
# Nested Comprehensions:

original_dict = {"a": 1, "b": 2, "c": 3}
swapped_dict = {value: key for key, value in original_dict.items()}
print("Swapped dictionary:", swapped_dict)
```

```python
# 8. Implement a nested list comprehension to generate a list of all pairs of numbers from two
# lists where the sum of the pair is even.

list_a = [1, 2, 3]
list_b = [4, 5, 6]
even_sum_pairs = [(x, y) for x in list_a for y in list_b if (x + y) % 2 == 0]
print("Pairs with an even sum:", even_sum_pairs)
```

```python
# 9. Create a list of lists using nested list comprehensions to generate a multiplication table up to
# 10x10.

multiplication_table = [[i * j for j in range(1, 11)] for i in range(1, 11)]
print("Multiplication Table:")
for row in multiplication_table:
    print(row)
```

```python
# 10. Write a nested dictionary comprehension to flatten a dictionary with nested dictionaries,
# where the keys of the inner dictionaries become keys in the resulting dictionary.
# Advanced Comprehensions:


nested_dict = {
    "group1": {"a": 1, "b": 2},
    "group2": {"c": 3, "d": 4}
}
flattened_dict = {inner_key: inner_value 
                  for outer_key in nested_dict 
                  for inner_key, inner_value in nested_dict[outer_key].items()}
print("Flattened dictionary:", flattened_dict)
```

```python
# 11. Use a list comprehension with a conditional expression to create a list of numbers that are
# multiples of 3 but not multiples of 2 from 1 to 50.

multiples_condition = [i for i in range(1, 51) if i % 3 == 0 and i % 2 != 0]
print("Multiples of 3 (but not 2):", multiples_condition)
```

```python
# 12. Implement a dictionary comprehension to filter and transform a dictionary, keeping only the
# key-value pairs where the values are greater than a certain threshold and subtracting the
# threshold from each value.

data = {"a": 12, "b": 8, "c": 15, "d": 10}
threshold = 10
filtered_transformed = {k: v - threshold for k, v in data.items() if v > threshold}
print("Filtered and transformed dictionary:", filtered_transformed)
```

```python
# 13. Create a list comprehension that generates all prime numbers up to a given limit using the
# Sieve of Eratosthenes algorithm

limit = 50
sieve = [True] * (limit + 1)
sieve[0] = sieve[1] = False  # 0 and 1 are not prime

# Mark the non-prime numbers.
for p in range(2, int(limit ** 0.5) + 1):
    if sieve[p]:
        for i in range(p * p, limit + 1, p):
            sieve[i] = False

# List comprehension to get the prime numbers.
primes = [i for i, is_prime in enumerate(sieve) if is_prime]
print("Primes up to", limit, ":", primes)
```
