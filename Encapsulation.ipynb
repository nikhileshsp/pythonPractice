```python
# using public member

class Employee:
    def __init__(self,name,salary):

        self.name=name
        self.salary= salary
    
    def show(self):
        print("NAme: ", self.name,"salary: ",self.salary)

# creating object of a class
emp= Employee("Minal",48500)

# accessing public data member
print("name :",emp.name,"salary: ",emp.salary)

# claaing public method of the class
emp.show()
```

```python
# using private member

class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.__salary=salary  # it is private 

    def show(self):
        # private member are accessible from a class
        print("name: ",self.name,"Salary: ",self.__salary)

# creating object of a class

emp= Employee("Pratham",40025)

# accessing private data members
# print('Salary :',emp.__salary)

emp.show()
```

```python
# name mangling to access private members
class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.__salary=salary

# creating object of a class
emp= Employee('Nikhil',35000)
print('Name: ',emp.name)

# direct access to private member using name mangling
print('Salary',emp._Employee__salary)
```

```python
class Company:
    def __init__(self):
        self._project= "Raisoni Collage"

class Employee(Company):
    def __init__(self,name):
        self.name=name
        Company.__init__(self)

    def show(self):
        print("Employee name: ",self.name)
        # accessing protected member in child class
        print("working on project:",self._project)

c= Employee("Adarsh")
c.show()
# Direct access protected data member
print('Project:',c._project)
```

```python
class BankAccount:
    def __init__(self,account_number,balance):
        self.account_number= account_number     #public
        self._account_holder= "Priya Sharma"    # protected
        self.__balance= balance                 #private
    def deposite(self,amount):
        if amount>0:
            self.__balance+=amount
            print(f"Deposited Rs.{amount}. New Balance is Rs. {self.__balance}")
        else:
            print("Deposited amount must be positive")

    def withdraw(self,amount):
        if amount>0 and amount<=self.__balance:
            self.__balance-=amount
            print(f"Withdraw Rs.{amount}. New Balance is Rs.{self.__balance}.")
        else:
            print("Insufficient Funds or invalid amount")

    def mybalance(self):
        return self.__balance

b= BankAccount("123456789",50000)
print(f"Account Number: {b.account_number}")
print(f"Account Holder: {b._account_holder}")
print(f"Balance: Rs. {b.mybalance()}")

# perform some operations
b.deposite(50000)
b.withdraw(200000)

# Display account details after operations
print(f"Balabce: Rs.{b.mybalance()}")
```
