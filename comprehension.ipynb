```python
# list without comprehension
list =[1,2,3,4,5,5,6,7,8,8,9,10]

output_list=[]

for var in list:
    if var %2==0:
     output_list.append(var)
print(output_list)
```

```python
list =[1,2,3,4,5,5,6,7,8,8,9,10]

using_list_comp=[var for var in list if var %2==0]
print(using_list_comp)
```

```python
a= [1,2,3]
b=[4,5,6]

c=[]
for x in a:
    for y in b:
        if (x+y)%2==0:
            c.append((x,y))
            
print(c)
```
