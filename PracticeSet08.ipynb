```python
# 1. Set Creation: Create two sets containing the names of students in two different classes.
# Print both sets.

class_a = {"Alice", "Bob", "Charlie"}
class_b = {"David", "Eve", "Alice"}
print(f"Class A: {class_a}")
print(f"Class B: {class_b}")
```

```python
# 2. Set Operations: Perform set operations like union, intersection, and difference between the
# two sets of student names. Print the results.


class_a = {"Alice", "Bob", "Charlie"}
class_b = {"David", "Eve", "Alice"}
union_set = class_a.union(class_b)  # Combine all unique elements
intersection_set = class_a.intersection(class_b)  # Common elements
difference_set = class_a.difference(class_b)  # Elements in class_a but not in class_b
print(f"Union: {union_set}")
print(f"Intersection: {intersection_set}")
print(f"Difference: {difference_set}")
```

```python
# 3. Set Additions and Removals: Create a set of colors. Add a new color, remove an existing
# color, and print the updated set.

# Create a set of colors and modify it
colors = {"Red", "Blue", "Green"}
colors.add("Yellow")  # Add new color
colors.remove("Blue")  # Remove existing color
print(f"Updated colors set: {colors}")
```

```python
# 4. Set Comprehension: Generate a set of squares for numbers from 1 to 10 using set
# comprehension.

# Generate a set of squares for numbers from 1 to 10
squares = {num**2 for num in range(1, 11)}
print(f"Squares: {squares}")
```

```python
# 5. Set Membership: Write a program that checks if a given element is present in a set. Print
# "Element found" or "Element not found" accordingly.

# Check if an element is present in the set
sample_set = {"Apple", "Banana", "Cherry"}
element = "Banna"
if element in sample_set:
    print("Element found")
else:
    print("Element not found")
```

```python
# 6. Dictionary Creation: Create a dictionary that maps the names of countries to their capital
# cities. Print the dictionary.

country_capitals = {"India": "New Delhi", "USA": "Washington, D.C.", "France": "Paris", "Japan": "Tokyo"}
print(country_capitals)
```

```python
# 7. Accessing Dictionary Values: Given a dictionary of country capitals, ask the user to enter a
# country name and print the corresponding capital. Handle the case where the country is not
# found.

country_capitals = {"India": "New Delhi", "USA": "Washington, D.C.", "France": "Paris", "Japan": "Tokyo"}
country = input("Enter a country name: ")
capital = country_capitals.get(country, "Country not found")  # Handle missing keys
print(f"Capital: {capital}")
```

```python
# 8. Dictionary Manipulation: Create a dictionary of student grades. Add a new student and their
# grade, and then remove a student. Print the updated dictionary.

student_grades = {"Alice": 85, "Bob": 90, "Charlie": 78}
student_grades["David"] = 88  # Add a new student
del student_grades["Alice"]  # Remove a student
print(student_grades)
```

```python
# 9. Dictionary Iteration: Iterate through the dictionary of country capitals and print all countrycapital pairs.

# Iterate through country capitals dictionary
country_capitals = {"India": "New Delhi", "USA": "Washington, D.C.", "France": "Paris", "Japan": "Tokyo"}
for country, capital in country_capitals.items():
    print(f"{country}: {capital}")

print(country_capitals)
```

```python
# 10. Dictionary Comprehension: Generate a dictionary that maps numbers to their squares for
# numbers from 1 to 10 using dictionary comprehension.
# Generate a dictionary mapping numbers to their squares


number_squares = {num: num**2 for num in range(1, 11)}
print(number_squares)
```

```python
# 11. Dictionary Key Validation: Write a program that checks if a given key exists in a dictionary.
# Print "Key exists" or "Key does not exist" accordingly.

# Check if a key exists in a dictionary
student_grades = {"Alice": 85, "Bob": 90, "Charlie": 78}
key_to_check = "Bob"
if key_to_check in student_grades:
    print("Key exists")
else:
    print("Key does not exist")
```

```python
# 12. Dictionary Merging: Create two dictionaries representing the scores of two students. Merge
# these dictionaries into a single dictionary and print the result.

# Merge two dictionaries
scores_student1 = {"Math": 85, "Science": 90}
scores_student2 = {"English": 78, "History": 88}
merged_scores = {**scores_student1, **scores_student2}  # Merge dictionaries
print(merged_scores)
```

```python
# 13. Nested Dictionaries: Create a dictionary that represents a library catalog. Each book is
# identified by an ISBN (a unique number) and contains information such as the title, author,
# and publication year.

# Create a library catalog using nested dictionaries
library_catalog = {
    "978-3-16-148410-0": {"Title": "Python Basics", "Author": "John Doe", "Year": 2022},
    "978-1-4028-9462-6": {"Title": "Advanced Python", "Author": "Jane Smith", "Year": 2023}
}
print(library_catalog)
```

```python
# 14. Dictionary Sorting: Sort the keys of a dictionary alphabetically and print the sorted keys.

# Sort dictionary keys alphabetically
student_grades = {"Charlie": 78, "Alice": 85, "Bob": 90}
sorted_keys = sorted(student_grades.keys())
print(sorted_keys)
```
