```python
#  Doing indexing  

a= 'PYTHON'
print(a[0])     #  using normal indexing to print first letter
print(a[-6])     # using negative indexion to print first letter 
print(a[5])     #  using normal indexing to print last indexing
print(a[-1])     # using negative indexing to print last letter
```

```python
# permorming sliceing

a= 'PYTHON'
print(a[0:])      # starting at 0
print(a[0:2])      # Stoping aat 2
print(a[0:3])     # Starting at 0and stoping at 3
print(a[0:5:2])   # Starting at 0, stoping at 5 and taking step of 2
print(a[:])       # not starting or stoping anywhere
```

```python
# performing sliceing
a= 'ELEPHANT'
print(a[0::7])
print(a[-8::7])

print(a[0:7:3])
print(a[-8:-1:3])

print(a[-8:-5:2])
print(a[0:3:2])

print(a[5:])
print(a[-3:])


print(a[0:7:2])
print(a[-8:-1:2])
print(a[-8::2])

```
