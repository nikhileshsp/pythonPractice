```python
# using write and read method
file =open("myfile.txt","w")
file.write("This is the write command")
file.write("\n allows us to write in a particular file")
file.close()
# to read file now ,we need to open the file again
file= open("myfile.txt","r")
print(file.read())
```

```python
# using write and read method
file =open("myfile.txt","w+")
file.write("This is the write command")
file.write("\n allows us to write in a particular file")
#file.close()
# to read file now ,we need to open the file again
#file= open("myfile.txt","r")
print(file.tell())
file.seek(0)
print(file.read())
print(file.tell())
```
