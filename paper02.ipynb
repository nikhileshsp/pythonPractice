```python
#02

num= int(input("Enter a number"))
sum_of=0
while num>0:
    digit= num %10
    sum_of +=digit
    num=num//10
print("the sum of digit:",sum_of)
```

```python
#03

num= int(input("Enter the number"))
if num>0:
    if num %2==0:
        print("positive Even Number")
    else:
        print("positive odd number")
elif num<0:
    if num%2==0:
        print("Negative even number")
    else:
        print("negative odd number")
else:
    print("zero")
```
