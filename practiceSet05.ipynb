```python
# 03
# Elif Statement: Write a program that takes a numerical grade as input and converts it to a
# letter grade based on the following scale:

grade = float(input("Enter your numerical grade: "))

# Convert numerical grade to letter grade
if grade >= 90:
    print("Letter Grade: A")
elif grade >= 80:
    print("Letter Grade: B")
elif grade >= 70:
    print("Letter Grade: C")
elif grade >= 60:
    print("Letter Grade: D")
else:
    print("Letter Grade: F")
```

```python
# 04
# s: Create a program that asks the user for two numbers and then checks
# whether the first number is divisible by the second number. If it is, print "Divisible," otherwise,
# print "Not divisible."

num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

# Check divisibility
if num2 != 0:  # Ensure the second number is not zero
    if num1 % num2 == 0:
        print("Divisible")
    else:
        print("Not divisible")
else:
    print("Division by zero is not allowed!")
```

```python
# 05.1
# Write a program that prompts the user for their age and whether they have a driver's
# license. If the user is 18 or older and has a driver's license, print "You are eligible to
# drive." Otherwise, print "You are not eligible to drive."

age = int(input("Enter your age: "))
has_license = input("Do you have a driver's license? (yes/no): ").strip().lower()

if age >= 18 and has_license == "yes":
    print("You are eligible to drive.")
else:
    print("You are not eligible to drive.")
```

```python
# 07
# Password Validation: Write a program that asks the user to enter a password. If the password
# is less than 8 characters long, print "Password is too short." If it contains both letters and
# numbers, print "Password is valid." Otherwise, print "Password is invalid."

password = input("Enter a password: ")

if len(password) < 8:
    print("Password is too short.")
elif any(char.isalpha() for char in password) and any(char.isdigit() for char in password):
    print("Password is valid.")
else:
    print("Password is invalid.")
```

```python
# 08
# Leap Year Checker: Create a program that checks if a given year is a leap year. A leap year is
# divisible by 4, but not divisible by 100 unless it is also divisible by 400. Print either "Leap year"
# or "Not a leap year."

year = int(input("Enter a year: "))

# Check if it's a leap year
if year % 4 == 0:
    if year % 100 != 0 or year % 400 == 0:
        print("Leap year")
    else:
        print("Not a leap year")
else:
    print("Not a leap year")
```

```python
# 09
# BMI Calculator: Write a program that calculates a person's BMI (Body Mass Index) based on
# their weight (in kilograms) and height (in meters). Then, categorize their BMI as Underweight
# (<18.5), Normal (18.5-24.9), Overweight (25-29.9), or Obese (30 or greater).


# Input weight and height from the user
weight = float(input("Enter your weight (in kilograms): "))
height = float(input("Enter your height (in meters): "))

# Calculate BMI
bmi = weight / (height ** 2)

# Print BMI and categorize it
print(f"Your BMI is: {bmi:.2f}")

if bmi < 18.5:
    print("Category: Underweight")
elif 18.5 <= bmi <= 24.9:
    print("Category: Normal")
elif 25 <= bmi <= 29.9:
    print("Category: Overweight")
else:
    print("Category: Obese")
```

```python
# 10
# Logical Operators: Write a Python program that checks if a given number is both even and
# positive. If both conditions are met, it should print "The number is even and positive."

num = int(input("Enter a number: "))

# Check if the number is even and positive
if num > 0 and num % 2 == 0:
    print("The number is even and positive.")
else:
    print("The number does not meet the criteria.")
```

```python
# 11
# Short-circuiting: Create a program that checks if a user's input is a number within the range
# of 1 to 100 (inclusive). If it's not a number or it's outside the range, print an error message.


user_input = input("Enter a number: ")

# Check if the input is a number and within the range
if user_input.isdigit() and 1 <= int(user_input) <= 100:
    print("Valid input!")
else:
    print("Error: Input is either not a number or out of range.")
```

```python
# 12
# Multiple Conditions with elif: Develop a program that takes the current temperature as input
# and provides recommendations like "Wear a coat" for temperatures below freezing, "Wear a
# jacket" for temperatures between 0 and 10 degrees Celsius, and "Wear a t-shirt" for
# temperatures above 10 degrees Celsius.

temperature = float(input("Enter the current temperature (in Celsius): "))

if temperature < 0:
    print("Recommendation: Wear a coat.")
elif 0 <= temperature <= 10:
    print("Recommendation: Wear a jacket.")
else:
    print("Recommendation: Wear a t-shirt.")
```
