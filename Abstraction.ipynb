```python
from abc import ABC
class Car(ABC):
    def mileage(self):
        pass

class Tesla(Car):
    def mileage(self):
        print("the mileage is 30kmpl")

class Suzuki(Car):
    def mileage(self):
        print("the mileage is 25kmpl")

class Duster(Car):
    def mileage(self):
        print("the mileage is 24kmpl")

# creating object
t= Tesla()
s= Suzuki()
d= Duster()
t.mileage()
s.mileage()
d.mileage()
```

```python
from abc import ABC

class Media(ABC):
    def get_info(self):
        pass

    def play(self):
        pass

class Book(Media):
    def __init__(self,title,author):
        self.title= title
        self.author= author

    def get_info(self):
        return f"Book : {self.title} by {self.author}"

    def play(self):
        print(f"Reading '{self.title}'By {self.author}")
        

class Movie(Media):
    def __init__(self,title,director):
        self.title= title
        self.director= director

    def get_info(self):
        return f"Movie : {self.title} by {self.director}"

    def play(self):
        print(f"Playing '{self.title}'By {self.director}")

b= Book("1984","George Orwell")
m= Movie("Inception","Christopher Nolan")

print(b.get_info())
b.play()
print(m.get_info())
m.play()
```
