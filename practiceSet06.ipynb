```python
# 01
# Basic Looping:
#  Write a for loop to print all the numbers from 1 to 10.
#  Write a while loop to print all the even numbers from 2 to 20.
#  Write a for loop to print sum of all even numbers and sum of all odd numbers from 1
# to 100.

for num in range(1, 11):
    print(num)
# ------------------------------
num = 2
while num <= 20:
    print(num)
    num += 2

# --------------------------------
sum_even = 0
sum_odd = 0

for num in range(1, 101):
    if num % 2 == 0:
        sum_even += num
    else:
        sum_odd += num

print("Sum of all even numbers from 1 to 100:", sum_even)
print("Sum of all odd numbers from 1 to 100:", sum_odd)
```

```python
# 02
# Loop Control:
#  Write a program that sums all the numbers from 1 to 100 using a for loop.
#  Create a while loop that counts down from 10 to 1 and then prints "Blastoff!".

total = 0
for num in range(1, 101):
    total += num
print("The sum of all numbers from 1 to 100 is:", total)

# ---------------------------------------------------------------
count = 10
while count > 0:
    print(count)
    count -= 1
print("Blastoff!")
```

```python
# 03
# Looping with Lists:
#  Given a list of numbers, use a for loop to find the sum of all the numbers in the list,
# where list=[1,4,5,6,7,4,1,10]
#  Write a for loop to print each element of a list of fruits, where
# a=[‘apple’,’banana’,’strawberry’,blueberry’,’pineapple’]


numbers = [1, 4, 5, 6, 7, 4, 1, 10]
total = 0
for num in numbers:
    total += num

print("The sum of all numbers in the list is:", total)

# ------------------------------------------------------------
# List of fruits
fruits = ['apple', 'banana', 'strawberry', 'blueberry', 'pineapple']

# Use a for loop 
for fruit in fruits:
    print(fruit)
```

```python
# 04
# Nested Loops:
#  Use nested for loops to print a multiplication table for numbers from 1 to 10.
#  Write a program that generates a pattern of asterisks in the shape of a triangle

for i in range(1, 11):
    for j in range(1, 11):
        print(f"{i}x{j} = {i*j}\t",end=' ')
        # print(" ")
    print()  # Blank line for better readabilit\y
# -------------------------------------------------------------------------------
rows = int(input("Enter the number of rows for the triangle: "))

for i in range(1, rows + 1):
    print("*" * i)
```

```python
# 05
#   Looping with Strings:
#  Write a for loop to count the number of vowels (a, e, i, o, u) in a given string.
#  Use a while loop to reverse a string.

text = input("Enter a string: ")

vowel_count = 0

vowels = "aeiouAEIOU"
for char in text:
    if char in vowels:
        vowel_count += 1

print("Number of vowels in the string:", vowel_count)

# -------------------------------------------------------------------

text = input("Enter a string: ")

reversed_text = ""

index = len(text) - 1

while index >= 0:
    reversed_text += text[index]
    index -= 1

print("Reversed string:", reversed_text)
```

```python
06
Loop Control Statements:
 Write a for loop to find the first prime number in a given range.
 Create a program that uses a for loop and the break statement to find and print the
smallest number in a list.


start = int(input("Enter the start of the range: "))
end = int(input("Enter the end of the range: "))

for num in range(start, end + 1):
    if num > 1:  
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:  # Not a prime number
                break
        else:
            print("The first prime number in the range is:", num)
            break
else:
    print("No prime number found in the given range.")

# ----------------------------------------------------

# List of numbers
numbers = [45, 3, 29, 18, 4, 10]

# Find the smallest number
smallest = numbers[0]  # Assume the first number is the smallest
for num in numbers:
    if num < smallest:
        smallest = num

print("The smallest number in the list is:", smallest)
```

```python
# 07
# Looping with Dictionaries:
#  Iterate through a dictionary of student names and their corresponding scores, and
# print out the names of students who scored above 90, where dictionary is
# mystudents={“riya”:45,”miya”:98,”siya”:97,”geeta”:91,”Mahesh”:89,”ganesh”:94}
#  Create a dictionary representing a shopping cart with items and their prices. Use a for
# loop to calculate the total cost of the items in the cart.


mystudents = {"riya": 45, "miya": 98, "siya": 97, "geeta": 91, "Mahesh": 89, "ganesh": 94}

# Iterate and print names of students who scored above 90
print("Students who scored above 90:")
for name, score in mystudents.items():
    if score > 90:
        print(name)

# -------------------------------------------------------

# Dictionary representing a shopping cart (item: price)
shopping_cart = {"apple": 30, "banana": 10, "milk": 50, "bread": 40, "eggs": 60}

# Calculate the total cost using a for loop
total_cost = 0
for item, price in shopping_cart.items():
    total_cost += price

print("Total cost of items in the shopping cart:", total_cost)

```

```python
# 08
# Infinite Loops and Loop Control:
#  Write a program with an infinite loop, and use the break statement to exit the loop
# when a specific condition is met.
#  Create a program that asks the user to guess a number between 1 and 10. Use a while
# loop to keep asking until they guess correctly.


# Infinite loop
while True:
    user_input = input("Enter 'exit' to break the loop: ").strip().lower()
    if user_input == "exit":
        print("Exiting the loop...")
        break
    else:
        print("You didn't type 'exit'. Try again!")

# --------------------------------------------

import random


correct_number = random.randint(1, 10)

# Guessing loop
while True:
    guess = int(input("Guess a number between 1 and 10: "))
    if guess == correct_number:
        print("Congratulations! You guessed it correctly.")
        break
    else:
        print("Incorrect guess. Try again!")

```

```python
# 09
# Looping:
#  Write a program to find the Fibonacci sequence up to a specified number of terms
# using a loop.
#  Implement a guessing game where the computer generates a random number, and
# the player has to guess it within a certain number of tries. Provide hints along the way.

terms = int(input("Enter the number of terms for the Fibonacci sequence: "))

# Initialize the first two terms
a, b = 0, 1

# Generate the Fibonacci sequence using a loop
print("Fibonacci sequence:")
for _ in range(terms):
    print(a, end=" ")
    a, b = b, a + b

# -------------------------------------------------

import random

# Generate a random number between 1 and 100
target_number = random.randint(1, 100)
max_tries = 10
attempt = 0

print("Guess the number between 1 and 100. You have 10 tries!")

while attempt < max_tries:
    guess = int(input(f"Attempt {attempt + 1}: Enter your guess: "))
    attempt += 1

    if guess == target_number:
        print("Congratulations! You guessed the number correctly!")
        break
    elif guess < target_number:
        print("Hint: The number is higher.")
    else:
        print("Hint: The number is lower.")

if attempt == max_tries and guess != target_number:
    print(f"Sorry, you've used all your tries. The correct number was {target_number}.")
```
