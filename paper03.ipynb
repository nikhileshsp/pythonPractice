```python
# 01
# Create a class called `Book` with two attributes: `title` and `author`. 
# Then create an object of this class with the title 
# "Harry Potter" and author "J.K. Rowling", and print the title and author.

class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
    def display_info(self):
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")
book1 = Book("Harry Potter", "J.K. Rowling")

book1.display_info()


```

```python
# 02
# Create a class called `Car` that has a constructor (`__init__`) which takes `brand` and `model` as parameters.
# When an object is created, it should print: "Car created: <brand> <model>"

class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
        print(f"Car created:{brand} {model}")

car1 = Car("tata", "altroz")
```

```python
# 03
# Create a base class `Animal` with a method `speak()` that prints "Animal speaks".
# Then create a derived class `Dog` that overrides the `speak()`
# method to print "Dog barks". Create an object of `Dog` and call its `speak()` method.


class Animal:
    def speak(self):
        print("Animal speaks")
        
class Dog(Animal):
    def speak(self):
        print("Dog barks")


dog1 = Dog()
dog1.speak()
```

```python
# 04
# Create three classes:
# - `Animal` with method `sound()` → prints "Animals make sounds"
# - `Mammal` inherits from `Animal` and adds method `has_hair()` → prints "Most mammals have hair"
# - `Dog` inherits from `Mammal` and adds method `bark()` → prints "Dog barks"
# Create an object of `Dog` and call all three methods.

# Base class
class Animal:
    def sound(self):
        print("Animals make sounds")

class Mammal(Animal):
    def has_hair(self):
        print("Most mammals have hair")

class Dog(Mammal):
    def bark(self):
        print("Dog barks")
dog1 = Dog()


dog1.sound()     
dog1.has_hair() 
dog1.bark()      

```
