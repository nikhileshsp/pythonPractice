```python
def square(x):
    return x**2
number=[1,2,3,4]
squared_number=map(square,number)
print(list(squared_number))
```

```python
# same program using comprehension and loop

def square(x):
    return x **2
number=[1,2,3,4]
squared_number=[square(x) for x in number]
print(squared_number)
```

```python
# same program using comprehension and loop
def square(x):
    return x**2
numbers=[1,2,3,4]
squared_numbers=[]
for x in numbers:
    squared_numbers.append(square(x))
print(squared_numbers)
    
```

```python
def multiply_by(factor):
    def multiplier(x):
        return x *factor
    return multiplier
double= multiply_by(2)
tripal= multiply_by(3)
print(double(5))
print(tripal(5))
```

```python
def myfunction(x):
    def multiplication(y):
        return y*2
    return multiplication(x)+3   
    
result= myfunction(5)


# innner_function i.e multiplication(5) return 10,
# then adds 3 by outer_function i.e myfunction
print(result)
```

```python
def longest_value(d:dict):
    longest= max(d.values(),key=len)
    return longest

fruits={'fruits':'apple','colour':'green'}
print(longest_value(fruits))
```
