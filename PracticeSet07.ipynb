```python
# 1. List Creation: Create a list containing the names of your favorite fruits. Print the list.

# Create a list of favorite fruits
fruits = ["apple", "banana", "mango", "orange"]
print("Favorite fruits:", fruits)
```

```python
# 2. List Manipulation: Add a new fruit to the list and print the updated list. Then, remove one of

# Add a new fruit
fruits = ["apple", "banana", "mango", "orange"]
fruits.append("grape")
print("After adding a fruit:", fruits)

# Remove a fruit
fruits.remove("banana")
print("After removing a fruit:", fruits)
```

```python
# 3. List Slicing: Take a list of numbers and print the first three elements, the last three elements,
# and a subset of elements in the middle.

# List of numbers
numbers = [10, 20, 30, 40, 50, 60, 70, 80]

# Slicing the list
print("First three elements:", numbers[:3])
print("Last three elements:", numbers[-3:])
print("Subset of elements in the middle:", numbers[2:6])

```

```python
# 4. List Comprehension: Generate a list of squares for numbers from 1 to 10 using list
# comprehension.

squares = [x**2 for x in range(1, 11)]
print("List of squares:", squares)

```

```python
# 5. List Sorting: Create a list of unsorted numbers and then sort them in ascending order. Print
# both the unsorted and sorted lists.

# Create an unsorted list
unsorted_numbers = [40, 10, 70, 20, 50]
print("Unsorted list:", unsorted_numbers)

# Sort the list
unsorted_numbers.sort()
print("Sorted list:", unsorted_numbers)
```

```python
# 6. List Concatenation: Combine two lists into a single list and print the result.

# Combine two lists
list1 = [1, 2, 3]
list2 = [4, 5, 6]
combined_list = list1 + list2
print("Combined list:", combined_list)
```

```python
# 7. List Searching: Write a program that checks if a given item is present in a list. Print "Item
# found" or "Item not found" accordingly.

# Check if an item is in the list
items = ["pen", "book", "laptop"]
item_to_search = "book"

if item_to_search in items:
    print("Item found")
else:
    print("Item not found")
```

```python
# 8. List Removal: Create a list of numbers and remove all occurrences of a specific number from
# the list. Print the updated list.

numbers = [1, 2, 3, 2, 4, 2, 5]
number_to_remove = 2

# Remove using list comprehension
updated_list = [num for num in numbers if num != number_to_remove]
print("Updated list after removal:", updated_list)
```

```python
# 9. Tuple Creation: Define a tuple containing the days of the week. Print the tuple.

days_of_week = ("Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday")
print(days_of_week)
```

```python
# 10. Tuple Unpacking: Create a tuple that contains your name and age. Use tuple unpacking to
# assign these values to two variables and print them.

# Create a tuple with name and age
personal_info = ("Nikhilesh", 25)  # Replace with your name and age
name, age = personal_info  # Unpacking the tuple
print(f"Name: {name}, Age: {age}")
```

```python
# 11. Tuple Concatenation: Combine two tuples into a single tuple and print the result.

tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
combined_tuple = tuple1 + tuple2
print(combined_tuple)
```

```python
# 12. Tuple as Dictionary Key: Create a dictionary where the keys are tuples (e.g., (1, 2)) and the
# values are corresponding descriptions. Access a value using a tuple as the key.

tuple_dict = {
    (1, 2): "First pair",
    (3, 4): "Second pair"
}
# Access a value using a tuple key
print(tuple_dict[(1, 2)])  # Output: First pair
print(tuple_dict[(3, 4)])  # Output: second pair
```

```python
# 13. Tuple Immutability: Attempt to modify a tuple by changing one of its elements. Explain the
# error that occurs.

# Attempt to modify a tuple
immutable_tuple = (1, 2, 3)

immutable_tuple[0] = 10  # This will raise an error
```

```python
# 14. Tuple Conversion: Convert a tuple into a list, make changes to the list, and then convert it
# back into a tuple. Print the tuple before and after the conversion.

# Convert a tuple into a list, modify it, and convert back
convert_tuple = (10, 20, 30)
convert_list = list(convert_tuple)  # Convert to list
convert_list[1] = 25  # Modify the list
convert_back = tuple(convert_list)  # Convert back to tuple
print(f"Before: {convert_tuple}, After: {convert_back}")
```

```python
# 15. Tuple Comparison: Compare two tuples element-wise and print whether they are equal or
# not.

# Compare two tuples element-wise
tuple_a = (1, 2, 3)
tuple_b = (1, 2, 8)
print(f"Are the tuples equal? {tuple_a == tuple_b}")  # Output: True
```
