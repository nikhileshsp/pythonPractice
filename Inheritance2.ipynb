```python
# example of single inheritance
# base \parentclass
class Vehicle:
    def __init__(self,companyname,year):
        self.companyname= companyname
        self.year=year
    def Vehicle_info(self):
        print('My Vehicle Details are ',self.companyname,self.year)

class Car(Vehicle):
    def __init__(self,companyname,year,carcolor,purchasedate):
        super().__init__(companyname,year)
        self.carcolor=carcolor
        self.purchasedate=purchasedate
    def car_info(self):
        print('My car Details are',self.carcolor,self.purchasedate)

accent= Car("Hyundai",2019,"Red","19th jul 2022")
accent.Vehicle_info()
accent.car_info()
```

```python
# default function

def greet(name="Guest"):
    return f"Hello, {name}!"

# Calling function without argument (uses default value)
print(greet())  # Output: Hello, Guest!

# Calling function with argument
print(greet("Nikhilesh"))  # Output: Hello, Nikhilesh!
```
