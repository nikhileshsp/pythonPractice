```python
1. Array Basics
What is an array in Python?

An array is a data structure that stores elements of the same type in a contiguous block of memory. It is used for efficient numerical operations.

How are arrays different from lists in Python?

Arrays require all elements to be of the same type (e.g., integers, floats) and are more memory-efficient for numerical operations. Lists, on the other hand, can store elements of mixed types.

Create an array with elements 1, 2, 3, and 4 in Python:

python
import array
arr = array.array('i', [1, 2, 3, 4])  # 'i' denotes integer type
print(arr)
2. Array Indexing
Access the third element in an array:

python
print(arr[2])  # Indexing starts from 0, so the third element is at index 2.
What happens if you access an out-of-bounds index?

It raises an IndexError, indicating you’re trying to access an invalid position.

Print elements of an array in reverse order:

python
print(arr[::-1])  # Slicing to reverse the array.
3. Array Operations
Difference between appending and extending an array:

append(x) adds a single element x to the end of an array.

extend([x, y]) adds multiple elements from a list to the array.

Concatenate two arrays:

python
arr1 = array.array('i', [1, 2])
arr2 = array.array('i', [3, 4])
concatenated = arr1 + arr2
print(concatenated)
Find the length of an array:

python
print(len(arr))  # Returns the number of elements in the array.
4. Array Manipulation
Add elements 5 and 6 at the end of an array:

python
arr.extend([5, 6])  # Using extend to add multiple elements.
print(arr)
Remove the element at index 2:

python
arr.pop(2)
print(arr)
Replace the element at index 1 with value 10:

python
arr[1] = 10
print(arr)
5. Array Iteration
Iterate over each element and print its square:

python
for x in arr:
    print(x**2)
Check if a specific element exists in an array:

python
print(10 in arr)  # Returns True if 10 is in the array, False otherwise.
Create a new array by doubling each element:

python
doubled = array.array('i', [x * 2 for x in arr])
print(doubled)
6. Multidimensional Arrays
What is a multidimensional array?

A multidimensional array has rows and columns (like a matrix) and can store data in more than one dimension.

Create a 2x3 array with random integers:

python
import numpy as np
arr2d = np.random.randint(1, 10, (2, 3))  # 2 rows, 3 columns
print(arr2d)
Access the element at the second row and third column:

python
print(arr2d[1][2])  # Second row (index 1) and third column (index 2)
7. Array Slicing
Concept of array slicing:

Slicing extracts a portion of an array using a range of indices, e.g., arr[start:end:step].

Extract subarray from index 1 to 4:

python
subarray = arr[1:5]
print(subarray)
Reverse the elements using slicing:

python
reversed_arr = arr[::-1]
print(reversed_arr)
8. Array Functions
Find the maximum element:

python
print(max(arr))  # Built-in function.
Calculate the sum of all elements:

python
print(sum(arr))  # Built-in sum function.
Find the index of the first occurrence of a specific element:

python
print(arr.index(10))  # Index of the first occurrence of 10.
9. Array Sorting
Sort an array in ascending order:

python
arr_sorted = sorted(arr)  # Returns a sorted list.
print(arr_sorted)
Sort an array of strings alphabetically:

python
strings = ["banana", "apple", "cherry"]
strings.sort()
print(strings)
Custom sorting function (descending order):

python
arr_desc = sorted(arr, reverse=True)
print(arr_desc)
10. Array Challenges
Rotate elements of an array to the right by K positions:

python
def rotate_array(arr, k):
    k %= len(arr)
    return arr[-k:] + arr[:-k]

rotated = rotate_array(arr, 2)
print(rotated)
Find the intersection of two arrays:

python
arr1 = array.array('i', [1, 2, 3, 4])
arr2 = array.array('i', [3, 4, 5, 6])
intersection = [x for x in arr1 if x in arr2]
print(intersection)
Check if an array is a palindrome:

python
def is_palindrome(arr):
    return arr == arr[::-1]

print(is_palindrome(arr))  # True or False.
```
