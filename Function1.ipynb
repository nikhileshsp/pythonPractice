```python
# using default arguments
def greet(name,greeting="Hello"):
    print((f"{greeting},{name}!"))
 # calling with only required argument   
greet("Rashmi")
# calling with both required and default argument
greet("Rashmi","hi")
```

```python
def calculate_area(radius):
    '''
    we are calculate the area of a circle,
    parameters: the radius of the circle
    returns: float: the area of the circle
    '''
    pi = 3.1415
    area= pi*radius**2
    return area
a= calculate_area(4)
print("the area of the circle is:",a)
print(calculate_area.__doc__)
```
