```python
# using diffrent placeholder values named indexes
txt1= "my name is {fname}, i'am {age}".format(fname="rashmi",age=36)

# numbered indexes
txt2= " my name is {0}, i am {1}".format("rashmi",36)

# empty placeholder
txt3= " my name is {}, i am {}".format("rashmi",36)

print(txt1)
print(txt2)
print(txt3)
```
