```python
student_grade={
    'student':{'Python':90,'Data Science':85,'Data Analytics':92},
    'student2':{'Python':88,'Data Science':91,'Data Analytics':89}
}
for student,courses in student_grade.items():
    avg_grade= sum(courses.values())/len(courses)
    print(f"{student} average grade:{avg_grade}")
```
