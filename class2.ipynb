```python
#  Learning Logical 

a= 10
b= 20
print(a==b and a!=b)   # and operator
print(a<b or a>b)     # or operator
print(not(a<b and a>b))     # not with and operator  (Reverse output)
print(not(a<b or a>b))     # not with or operator    (Reverse output)
```

```python
# Learning Comparison Operator
a= 10
b=20
print(a==b) # Equal to
print(a!=b)  # not equal to
print(a<b)     # less then
print(a<=b)    # less then equal to 
print(a>b)     # greater then
print(a>=b)    # greater then equal to
```
