```python
def longest_value(d:dict):

    longest= max(d.values(),key=len)
    return longest
fruits={'fruit':'apple','color':'green'}
print(longest_value(fruits))
```

```python
def convert_add(a):
    b=[]
    for i in a:
        b.append(int(i))
    return sum(b)
print(convert_add(['1','3','5','6']))
```

```python
def check_duplicates(a):
    for x in a:
        # using cou to find items more then one
        if a.count(x)>1:
            return x
        else:
            return 'no duplicates'
fruits= ['apple','orange','banana','apple']
names=  ['Kuniket','rakesh','mrudul','pushpak','rakesh']
print(check_duplicates(fruits))
print(check_duplicates(names))
```

```python
def get_lowercase_sorted_tuple(names):
    d=[]
    for x in sorted(names,reverse= True):
        if x.islower():
            d.append(x)
    mytuple=tuple(d)
    return mytuple
names= ["kuniket","madhu","Shubham","Pallavi","pratham","Pushpak","rakesh"]
result= get_lowercase_sorted_tuple(names)
print(result)
```
