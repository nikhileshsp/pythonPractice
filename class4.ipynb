```python
 # reverse a string

original_str= "Hello World!"

reverse_str= (reversed(original_str))
print(reverse_str)
```

```python
# reverse a string

original_str= "Hello World!"

reverse_str= " ".join(reversed(original_str))
print(reverse_str)
```
