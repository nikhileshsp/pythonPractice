```python
class Vehicle:
    def __init__(self,name,color,price):
        self.name=name
        self.color=color
        self.price=price
    def show(self):
        print('Details: ',self.name,self.color,self.price)
    def max_speed(self):
        print('Vehicle max speed is 150')
    def change_gear(self):
        print('Vehicle change 6 gear')

class Car(Vehicle):
    def max_speed(self):
        print('Car max speed is 240')
    def change_gear(self):
        print('Car change 7 gear')

c1= Car('Car x1','Red',20000)
c1.show()

c1.max_speed()
c1.change_gear()

v1= Vehicle('Truck x1','white',75000)
v1.show()

v1.max_speed()
v1.change_gear()
```

```python
# ploymorphism in class methods
# creating parent class 01
class Hyundai:
    def fuel_type(self):
        print("patrol")
    def max_speed(self):
        print("Speed is 240")

class BMW:
    
    def fuel_type(self):
        print("diesel")
    def max_speed(self):
        print("speed is 350")

# creating both object of classses

h1= Hyundai()
b1= BMW()

for car in (h1,b1):
    car.fuel_type()
    car.max_speed()
```

```python
class Hyundai:
    def fuel_type(self):
        print("patrol")
    def max_speed(self):
        print("Speed is 350")

class BMW:
    
    def fuel_type(self):
        print("diesel")
    def max_speed(self):
        print("speed is 240")

def car_details(call):
    call.fuel_type()
    call.max_speed()

h1= Hyundai()
b1= BMW()

car_details(h1)
car_details(b1)
```

```python
# method overloading

class Maths:
    def sum(self,a,b):
        print("the sum is: ",a+b)
    def sum(self,a,b,c):
        print("the sum is: ",a+b+c)

m1= Maths()
# the below lines show an error
# m1.sum(2,25)

# this line will call the second product method
m1.sum(2,10,12)
```
